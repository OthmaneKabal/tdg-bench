```python
from pathlib import Path

from tdg_bench import TDGBench


# Fonctionne depuis la racine du projet ou depuis le dossier notebooks/
PROJECT_ROOT = next(
    path
    for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (path / "pyproject.toml").is_file()
)

DATA_ROOT = PROJECT_ROOT / "benchmark_data"

bench = TDGBench(DATA_ROOT)

result = bench.evaluate(
    dataset="biomed",
    graph="GT2KG_tdg",

    # Choix du GNN
    model="GCN",

    # Choix des embeddings
    embedding_model="sentence-transformers/all-MiniLM-L6-v2",

    split_seed=42,
    random_seed=42,

    device="auto",
    max_epochs=200,
    patience=100,

    # Afficher les métriques pendant l’apprentissage
    verbose=True,
    log_every=5,
)

print("\n===== RÉSULTAT FINAL =====")
print("Embedding :", result.embedding_model)
print("Epochs exécutées :", result.epochs_ran)
print("Meilleur epoch :", result.best_epoch)

print("\nTrain")
print("Loss :", result.train.loss)
print("Accuracy :", result.train.accuracy)
print("Macro-F1 :", result.train.macro_f1)

print("\nValidation")
print("Loss :", result.validation.loss)
print("Accuracy :", result.validation.accuracy)
print("Macro-F1 :", result.validation.macro_f1)

print("\nTest")
print("Loss :", result.test.loss)
print("Accuracy :", result.test.accuracy)
print("Macro-F1 :", result.test.macro_f1)
```

```python
from pathlib import Path
from tdg_bench import TDGBench

# Fonctionne même si le notebook est dans notebooks/
PROJECT_ROOT = next(
    path
    for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (path / "pyproject.toml").is_file()
)

bench = TDGBench(PROJECT_ROOT / "benchmark_data")

errors = []

print("Datasets disponibles :", bench.available_datasets())

for dataset in bench.available_datasets():
    print(f"\n{'=' * 70}")
    print(f"DATASET : {dataset}")
    print("Graphes :", bench.available_graphs(dataset))

    for graph in bench.available_graphs(dataset):
        print(f"\n--- Chargement de {dataset}/{graph} ---")

        try:
            bundle = bench.get_data(
                dataset=dataset,
                graph=graph,
                split_seed=42,
                initial_embedding_model=None,  # features aléatoires rapides
                feature_seed=42,
                verify_graph=True,
            )

            print("OK")
            print("Dataset :", bundle.dataset)
            print("Graphe :", bundle.graph)
            print("Split :", bundle.split_seed)
            print("Nombre de nœuds :", bundle.data.num_nodes)
            print("Nombre d'arêtes :", bundle.data.num_edges)
            print("Nombre de relations :", bundle.data.num_relations)
            print("Nombre de classes :", bundle.data.num_classes)
            print("Dimension des features :", bundle.data.x.shape)
            print("Couverture Gold Standard :", f"{bundle.coverage:.2%}")
            print("Tailles des splits :", bundle.split_sizes)
            print("Gold terms manquants :", len(bundle.missing_gold_terms))

        except Exception as error:
            errors.append((dataset, graph, error))
            print("ERREUR :", repr(error))

print(f"\n{'=' * 70}")

if errors:
    print(f"{len(errors)} graphe(s) en erreur :")
    for dataset, graph, error in errors:
        print(f"- {dataset}/{graph}: {error}")
    raise RuntimeError("Certains graphes n'ont pas pu être chargés.")

print("SUCCÈS : tous les graphes ont été chargés correctement.")
```
